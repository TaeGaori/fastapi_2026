```python
# 라이브러리 불러오기
import os
import pandas as pd
```

```python
BASE_DIR = os.getcwd()
BASE_DIR
```

```python
DATA_PATH = os.path.join(BASE_DIR, 'input', 'cctv_busan.csv')
DATA_PATH
```

```python
OUTPUT_PATH = os.path.join(BASE_DIR, 'output', 'cctv_busan_filter.csv')
OUTPUT_PATH
```

```python
df_raw = pd.read_csv(DATA_PATH, encoding='cp949')
df_raw.head()
```

```python
df_busan = df_raw[['관리기관명', '카메라대수', '설치연월']]
df_busan = df_busan.reset_index(drop=True)
```

```python
df_busan.head()
```

```python
df_busan.shape
```

```python
df_busan = df_raw[df_raw['관리기관명'].str.contains('부산')].copy()
df_busan['설치연도'] = df_busan['설치연월'].dropna().astype(str).str[:4]
result = df_busan.groupby('설치연도')['카메라대수'].sum().reset_index()
print(result)
```

```python
result['설치연도'] = result['설치연도'].astype(int)

df_busan_filter = result[(result['설치연도']>=2019)&(result['설치연도']<=2026)].copy()

df_busan_filter['누적카메라대수'] = result['카메라대수'].cumsum()

print(df_busan_filter)
```

```python
# CSV 저장

df_busan_filter.to_csv(OUTPUT_PATH, encoding='utf-8-sig', index=False)

print('csv 저장 완료!')
print(f'경로: {OUTPUT_PATH}')
print(f'크기: {len(df_busan_filter):,}행 x {len(df_busan_filter.columns)}열')
```
